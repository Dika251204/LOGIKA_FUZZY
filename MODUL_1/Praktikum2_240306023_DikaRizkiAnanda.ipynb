### Deskripsi Praktikum

Praktikum ini membahas pemodelan variabel linguistik menggunakan
logika fuzzy pada kasus penggunaan CPU server.

Semesta pembicaraan yang digunakan adalah U = [0, 100]%
beban CPU dengan tiga label linguistik, yaitu:

1. Rendah
2. Normal
3. Tinggi

Setiap label direpresentasikan menggunakan fungsi keanggotaan
fuzzy untuk mengetahui derajat keanggotaan suatu nilai penggunaan CPU.

## 1. Semesta Pembicaraan dan Label Linguistik

Semesta pembicaraan:

U = [0, 100]%

Label linguistik yang digunakan:

- **Rendah:** 0 ≤ x ≤ 40
  - Bahu kiri
  - Penuh pada x ≤ 20
  - Turun hingga 0 pada x = 40

- **Normal:** 30 ≤ x ≤ 70
  - Bentuk segitiga
  - Puncak pada x = 50

- **Tinggi:** 60 ≤ x ≤ 100
  - Bahu kanan
  - Mulai naik pada x = 60
  - Penuh pada x ≥ 80

## 2. Rumus Fungsi Keanggotaan

### A. Fungsi Keanggotaan Rendah

$$
\mu_{Rendah}(x)=
\begin{cases}
1, & x \le 20 \\
\dfrac{40-x}{40-20}, & 20 < x < 40 \\
0, & x \ge 40
\end{cases}
$$

### B. Fungsi Keanggotaan Normal

$$
\mu_{Normal}(x)=
\begin{cases}
\dfrac{x-30}{50-30}, & 30 \le x \le 50 \\
\dfrac{70-x}{70-50}, & 50 < x \le 70 \\
0, & \text{di luar interval tersebut}
\end{cases}
$$

### C. Fungsi Keanggotaan Tinggi

$$
\mu_{Tinggi}(x)=
\begin{cases}
0, & x \le 60 \\
\dfrac{x-60}{80-60}, & 60 < x < 80 \\
1, & x \ge 80
\end{cases}
$$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def cpu_rendah(x):
    """Fungsi keanggotaan CPU Rendah (bahu kiri)."""
    x = np.asarray(x, dtype=float)

    return np.where(
        x <= 20,
        1.0,
        np.where(
            x < 40,
            (40 - x) / (40 - 20),
            0.0
        )
    )
```

```python
def cpu_normal(x):
    """Fungsi keanggotaan CPU Normal (segitiga)."""
    x = np.asarray(x, dtype=float)

    naik = (x - 30) / (50 - 30)
    turun = (70 - x) / (70 - 50)

    return np.maximum(
        0.0,
        np.minimum(naik, turun)
    )
```

```python
def cpu_tinggi(x):
    """Fungsi keanggotaan CPU Tinggi (bahu kanan)."""
    x = np.asarray(x, dtype=float)

    return np.where(
        x <= 60,
        0.0,
        np.where(
            x < 80,
            (x - 60) / (80 - 60),
            1.0
        )
    )
```

## 3. Pengujian Fungsi Keanggotaan

Pengujian dilakukan untuk melihat derajat keanggotaan beberapa
nilai penggunaan CPU terhadap label Rendah, Normal, dan Tinggi.

```python
contoh_cpu = [10, 20, 30, 40, 50, 60, 70, 80, 90]

print(
    f"{'CPU (%)':<10} | "
    f"{'Rendah':<10} | "
    f"{'Normal':<10} | "
    f"{'Tinggi':<10}"
)

print("-" * 50)

for x in contoh_cpu:
    rendah = float(cpu_rendah(x))
    normal = float(cpu_normal(x))
    tinggi = float(cpu_tinggi(x))

    print(
        f"{x:<10} | "
        f"{rendah:<10.3f} | "
        f"{normal:<10.3f} | "
        f"{tinggi:<10.3f}"
    )
```

## 4. Fuzzifikasi Input CPU = 10%

Sesuai instruksi Praktikum 2, dilakukan pengujian fuzzifikasi
untuk input penggunaan CPU sebesar 10%.

```python
x_uji = 10

mu_rendah = float(cpu_rendah(x_uji))
mu_normal = float(cpu_normal(x_uji))
mu_tinggi = float(cpu_tinggi(x_uji))

print("Hasil fuzzifikasi CPU = 10%")
print(f"μRendah = {mu_rendah:.3f}")
print(f"μNormal = {mu_normal:.3f}")
print(f"μTinggi = {mu_tinggi:.3f}")
```

## 5. Visualisasi Fungsi Keanggotaan

Grafik berikut menampilkan tiga fungsi keanggotaan penggunaan CPU,
yaitu Rendah, Normal, dan Tinggi dalam satu grafik.

```python
x = np.linspace(0, 100, 501)

y_rendah = cpu_rendah(x)
y_normal = cpu_normal(x)
y_tinggi = cpu_tinggi(x)

plt.figure(figsize=(11, 6))

plt.plot(x, y_rendah, linewidth=2.5, label="Rendah")
plt.plot(x, y_normal, linewidth=2.5, label="Normal")
plt.plot(x, y_tinggi, linewidth=2.5, label="Tinggi")

plt.axvline(
    x=10,
    linestyle="--",
    alpha=0.6,
    label="Input pengujian = 10%"
)

plt.title(
    "Fungsi Keanggotaan Variabel Linguistik Penggunaan CPU Server",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Penggunaan CPU (%)", fontsize=11)
plt.ylabel("Derajat Keanggotaan", fontsize=11)

plt.xlim(0, 100)
plt.ylim(-0.05, 1.1)

plt.xticks(np.arange(0, 101, 10))
plt.yticks(np.arange(0, 1.1, 0.1))

plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(loc="best")

plt.tight_layout()

nama_file = "grafik_variabel_linguistik_cpu.png"

plt.savefig(
    nama_file,
    format="png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print(f"Grafik disimpan sebagai: {nama_file}")
```
